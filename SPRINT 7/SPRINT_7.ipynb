**Función:** Validamos los datos de cada empleado:
* Si nombre sin espacio contiene carácteres no alfabeticos, entonces lanza un error.
* Si teléfono contiene carácteres no numéricos, entonces lanza un error.
* Si experiencia tiene un valor negativo, lanza un error.
* Si català ,excel o estudis no es un número entre 0 y 3, lanza un error.

**Argumento:** los argumentos de la función son nombre, telefono, experience, catala, excell, estudis --> que es la información que tenemos separadas por comas en el listado de candidatos.

**Return**: Si ninguna condición anterior se cumple, retorno la tupla con nombre, telefono, experience, catala, excell, estudis.


```python
def validacion_dades(nombre, telefono, experience, catala, excell, estudis):

    if not nombre.replace(' ', '').isalpha():
        raise ValueError('Nombre no válido') # es una palabra clave que se utiliza para lanzar o generar una excepción (un error) de forma intencional.

    if not telefono.isdigit():
        raise ValueError('El teléfono sólo debe contener números')

    if experience < 0:
        raise ValueError('La experiencia debe ser un número positivo')

    if not 0 <= catala <= 3:
        raise ValueError('Catalán debe ser entre 0 y 3')

    if not 0 <= excell <= 3:
        raise ValueError('Excel debe ser entre 0 y 3')

    if not 0 <= estudis <= 3:
        raise ValueError('Estudios debe ser entre 0 y 3')

    return nombre, telefono, experience, catala, excell, estudis
```

```python
validacion_dades('Candidat Error', '611000000', 5, 3, 3, 2)
```

**Función**: calcula la puntucación que obtiene el candidato.

**Argumento**: valor de experiencia, catalán, excel y nivel de estudios

**Return**: puntación final

```python
def calcular_puntuacion(experience, catala, excell, estudis):
    return (experience * 1.5) + catala + excell + estudis
```

```python
calcular_puntuacion(5, 3, 3, 2)
```

**Función**. Clasifica a los candidatos según la puntuación conseguida.

**Argumento**: Cálculo de la puntuación

**Return**: categoría según su puntuación

```python
def classificar_candidat(puntuacion_final):

    if puntuacion_final >= 12:
        return 'Prioritat Alta'

    elif puntuacion_final >= 8:
        return 'Candidat adequat'

    else:
        return 'No prioritari'
```

```python
classificar_candidat(13.5)
```

**Función**: Por cada candidato dentro de la lista, imprime el listado de candidatos con la columna de *Clasificación*

**Argumento**: listado de los candidatos que han pasado todo el proceso de verificación.

**Return**: informe con los candidatos finales.

```python
def mostrar_informe(candidatos_finales):

    print('\n--- INFORME FINAL ---\n')

    for candidato in candidatos_finales:

        print(f"Nombre: {candidato['nombre']}")
        print(f"Teléfono: {candidato['telefono']}")
        print(f"Puntuación: {candidato['puntuacion_final']}")
        print(f"Clasificación: {candidato['clasificacion']}")
        print('-'* 40)
```

```python
mostrar_informe([
    {   'nombre': 'Laura Serra',
        'telefono': '608112233',
        'puntuacion_final': 13.5,
        'clasificacion': 'Prioritat Alta'
    }
])
```

**Función**: junta todas las funciones anteriores. Además limita el listado a 10 candidatos validados.

**Argumento**: listado de candidatos

**Return**: informe con los candidatos finales.

```python
def procesar_candidatos(lista_candidatos):

    candidatos_validos = []

    for candidato in lista_candidatos:

        nombre, telefono, exp, catala, excell, estudis = candidato

        try:
            validacion_dades(nombre, telefono, exp, catala, excell, estudis)
        except ValueError as error:
            print(f"Error en {nombre}: {error}")
            continue

        if len(candidatos_validos) >= 10:
            print("Se ha alcanzado el límite de 10 candidatos")
            break

        puntuacion = calcular_puntuacion(exp, catala, excell, estudis)
        clasificacion = classificar_candidat(puntuacion)

        candidatos_validos.append({
            'nombre': nombre,
            'telefono': telefono,
            'puntuacion_final': puntuacion,
            'clasificacion': clasificacion
        })

    return candidatos_validos
```

```python
def main_seleccio(lista_candidatos):
    candidatos_finals = procesar_candidatos(lista_candidatos)
    mostrar_informe(candidatos_finals)

main_seleccio(candidatos)
```

Se crea un diccionario con los posibles comportamientos de los clientes.
La clave es el código que va del 0 al 4, en cada código se guarda el segmento, la campaña publicitaria que se activaría y el factor de comportamiento.

El factor edad da un valor a cada segmento de edad.

```python
COMPORTAMIENTOS = {
    0: {
        'segment': 'Sense interès',
        'descripcio': 'Navega però no afegeix productes',
        'missatge': '¡No te vayas con las manos vacías!',
        'factor': 0.10},
    1: {
        'segment': 'Interès baix',
        'descripcio': 'Afegeix productes però no arriba al checkout',
        'missatge': '¿Necesitas ayuda para elegir?',
        'factor': 0.25},
    2: {
        'segment': 'Risc moderat',
        'descripcio': 'Arriba al checkout però no paga',
        'missatge': '¡Casi es tuyo!',
        'factor': 0.50},
    3: {
        'segment': 'Risc alt',
        'descripcio': 'Cistella de compra abandonat fa menys de 24 h',
        'missatge': '¡Ups! Se te olvidó algo en tu carrito!',
        'factor': 0.75},
    4: {
        'segment': 'Risc crític',
        'descripcio': 'Cistella de compra abandonat fa més de 24 h',
        'missatge': '¡Tu carrito está a punto de caducar!',
        'factor': 0.90}
}

FACTOR_EDAD = {
    '18-25' : 1.20,
    '26-40' : 1.00,
    '41-60' : 0.85,
    '60+' : 0.70
}
```

**Función:** Recibe un código de comportamiento ,del 0 al 4, y devuelve el segmento, el mensaje publicitario y el factor de comportamiento correspondientes.

**Argumento**: código comportamiento, dle 0 al 4.

**Return**: devuelve las características del código en *dades*.

```python
def obtenir_segment_i_factor(codi_comportament):
    dades = COMPORTAMIENTOS.get(codi_comportament)
    if dades is None:
        return None, None, None, None
    return (
        dades['segment'],
        dades['descripcio'],
        dades['missatge'],
        dades['factor']
    )
```

```python
obtenir_segment_i_factor(4)
```

**Función**: Busca el grupo de edad dentro del diccionario.

**Argumento**: grupo de edad

**Return**: obtenemos el factor de edad.

```python
def obtenir_factor_edat(grup_edat):
  return FACTOR_EDAD[grup_edat]
```

```python
obtenir_factor_edat('26-40')
```

**función**: calcula el riesgo

**Argumento**: recibe el factor edad y comportamiento.

**Return**: devuelve el riesgo de cada cliente que abandona el carrito de compra.

```python
def calcular_risc(factor_comp, factor_edad):
  resultado = factor_comp * factor_edad
  return round(resultado,2)
```

```python
calcular_risc(0.75, 0.85)
```

**Función**: recorre la lista de clientes y para cada cliente añade según su edad y su código de comportamiento, el riesgo, el segmento, mensaje publicatario...

**Argumento**: listado de clientes

**Return**: devuelve el listado de clientes ya procesados.

```python
def processar_clientes_segmentacion(clientes):
  clientes_procesados = []

  for cliente in clientes:
    segment, descripcio, missatge, factor_comportamiento = obtenir_segment_i_factor(
        cliente['codi_comportament']
    )
    factor_edad = obtenir_factor_edat(
        cliente['grup_edat']
    )
    riesgo = calcular_risc(
        factor_comportamiento,
        factor_edad
    )
    nuevo_cliente = {
        'id_client':cliente['id_client'],
        'nom':cliente['nom'],
        'edat': cliente['edat'],
        'grup_edat':cliente['grup_edat'],
        'codi_comportament':cliente['codi_comportament'],
        'segment':segment,
        'mensaje_publicitario':missatge,
        'descripcio':descripcio,
        'factor_comportamiento':factor_comportamiento,
        'factor_edad':factor_edad,
        'riesgo':riesgo
    }
    clientes_procesados.append(nuevo_cliente)
  return clientes_procesados
```

```python
processar_clientes_segmentacion(clientes)
```

Diccionario con los niveles de riesgo de los sintomas de los clientes.

```python
nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

**Función**: Limpieza de datos.
* Se separa cada tipo de información usando el carácter '|'
* Con strip eliminanos los espacion del principio y del final.
* Se unifican los nombres de los sintomas para que todos tengan la misma sintaxis.
* Se quitan espacios y se sustituyen por barra baja, se ponen todos los síntomas en minúsculas.
* Con todo esto se crea una lista limpia de sintomas.

**Argumentos**: registro de paciente.

**Return**: devuelve registro limpio y homogéneo.

```python
def separar_info(registre):
  nom, edat_str, simptomes_str = registre.split("|")
  edat = int(edat_str.strip())
  simptomes = [simptoma.strip().lower().replace(' ','_') for simptoma in simptomes_str.split(',')]
  return {"nom": nom.strip(), "edat": edat, "simptomes": simptomes}
```

```python
pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]
separar_info(pacients[0])
```

**Función**: Recorre todo los clientes y aplica la función *separar_info*

**Argumentos**: registro de pacientes.

**Return**: devuelve diccionario limpio y homogéneo.

```python
def processar_pacients(pacients):
  pacients_procesats = [separar_info(paciente) for paciente in pacients]
  return pacients_procesats
```

```python
processar_pacients(pacients)
```

**Función**: calcula el riesgo total del paciente. Se parte desde 0 y se van sumando el coeficiente de cada síntoma.

**Argumentos**: cada síntoma y su nivel de riesgo.

**Return**: La suma de todos los síntomas del paciente.

```python
def risc_total(simptomes, nivells_risc):
  risc_total = 0
  for simptoma in simptomes:
    risc_total += nivells_risc[simptoma]
  return risc_total
```

```python
risc_total(["tos", "fatiga"], nivells_risc)
```

**Función**: recorre cada paciente, crea una nueva clave y guarda el resultado del calculo del riesgo.

**Argumentos**: Listado de pacientes procesados.

**Return**: Listado procesados con la columna con el cálculo del riesgo.


```python
def column_risc(pacients_procesats):
  for pacient in pacients_procesats:
    pacient["risc_total"] = risc_total(pacient["simptomes"], nivells_risc)
  return pacients_procesats
```

```python
pacients_procesats = processar_pacients(pacients)
```

```python
column_risc(pacients_procesats)
```

**Función**: Ordena los pacientes procesados por el riesgo de mayor riesgo a menos.

**Argumentos**: Listado de pacientes procesados.

**Return**: Listado de pacientes de mayor riesgo a menos.


```python
def ordenar_pacients(pacients_procesats):

    return sorted(
        pacients_procesats,
        key=lambda x: x["risc_total"],
        reverse=True
    )
```

```python
ordenar_pacients(pacients_procesats)
```

```python
import pandas as pd

def crear_dataframe(pacients_ordenats):
    df = pd.DataFrame(pacients_ordenats)
    return df
```

**Función**: Junta todas las funciones anteriores

**Argumentos**: Listado de pacientes.

**Return**: Listado de pacientes general y listado ordenado de mayor riesgo a menos.


```python
def main_pacints():

    pacients_processats = processar_pacients(pacients)
    pacients_processats = column_risc(pacients_processats)
    pacients_ordenats = ordenar_pacients(pacients_processats)
    df_pacients = crear_dataframe(pacients_ordenats)

    print('---DATAFRAME PACIENTES---\n')
    print(df_pacients)

    print('\n---LISTADO POR ORDEN DE RIESGO---\n')

    for pacient in pacients_ordenats:
        print(f"{pacient['nom']} - Risc: {pacient['risc_total']}")
```

```python
nivells_seguro = {
    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01},
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05},
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15},
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30}
}
```

**Función**: Busca el código en el diccionario.

**Argumentos**: Código

**Return**: Devuelve la información asociadad al *codi*


```python
def info_risc(codi):
  return nivells_seguro[codi]
```

```python
info_risc(0)
```

**Función**: Aplica la fórmula para calcular el riesgo.

**Argumentos**: Precio base y edad.

**Return**: Precio final

```python
def preu_final(preu_base, edat):
  return round(preu_base * (1 + edat/100),2)
```

```python
preu_final(120, 42)
```

**Función**: Lista vacía que guarda los clientes procesados.

**Argumentos**: listado de clientes.

**Return**: Devuelve la información asociadad al *codi*

```python
def processar_clients_seguro(clients):
  clients_processats = []

  for client in clients:

    codi = client['codi']
    info = info_risc(codi)
    preu = preu_final(info['preu_base'], client['edat'])

    nuevo_cliente = {
        'nom':client['nom'],
        'edat': client['edat'],
        'categoria':info['categoria'],
        'preu_base':info['preu_base'],
        'preu_final':preu,
        'probabilitat_fraude':info['fraude']
    }
    clients_processats.append(nuevo_cliente)
  return clients_processats
```

```python
processar_clients_seguro(clients)
```

**Función**: Recorre la lista nueva de clientes procesados

**Argumentos**: Listado de clientes procesados.

**Return**: Imprime la nueva lista

```python
def mostrar_dataframe(clients_processats):
  print('\n----CLIENTES----\n')
  for cliente in clients_processats:
    print(cliente)
```

```python
def ordenar_clients_preu(clients_processats):

    clients_ordenats = sorted(
        clients_processats,
        key=lambda x: x["preu_final"],
        reverse=True
    )
    return clients_ordenats
```

**Argumentos**: Clientes procesados

**Return**: Devuelve lista con clientes ordenados por coeficiente de fraude, de mayor a menor.

```python
def ordenar_clients_frau(clients_processats):

    clients_ordenats = sorted(
        clients_processats,
        key=lambda x: x['probabilitat_fraude'],
        reverse=True
    )

    return clients_ordenats
```

```python
ordenar_clients_frau(processar_clients_seguro(clients))
```

**Función**: Da respuestas a varias preguntas

**Argumentos**: Clientes procesados

**Return**: Imprime las respuestas de más probabilidad de fraude, precio más alto y que usuario/s tendriamos que revisar manualmente. En este caso, decidí que todos los usuarios con un coeficiente de riesgo **mayor al 15%** sería proritario revisarlos.

```python
def respostas(clients_processats):

    orden_frau = ordenar_clients_frau(clients_processats)
    orden_preu = ordenar_clients_preu(clients_processats)

    mes_probable = orden_frau[0]
    mes_car = orden_preu[0]

    print('\nQuin client té més probabilitat de frau?\n')
    print(f"El cliente con más probabilidad de fraude es {mes_probable['nom']}, "
          f"con una probabilidad de fraude de {mes_probable['probabilitat_fraude']*100} %")

    print('\nQuin té el preu final més alt?\n')
    print(f"El precio final más alto es {mes_car['preu_final']} €, del cliente {mes_car['nom']}")

    print('\nQui seria prioritari de revisar manualment?\n')

    for client in clients_processats:

        if client['probabilitat_fraude'] > 0.15:

              print(
                  client["nom"],
                  client['categoria'],
                  client["probabilitat_fraude"]
              )
```

```python
respostas(processar_clients_seguro(clients))
```

**Función**: Unifica las funciones anteriores.

**Argumentos**: Clientes.

**Return**: Listado de clientes, listado de clientes ordenados por mayor riesgo de fraude, listado de clientes con un precio mayor.

```python
def main_clients(clients):

  clients_processats = processar_clients_seguro(clients)
  mostrar_dataframe(clients_processats)
  respostas(clients_processats)
```

```python
main_clients(clients)
```

**Función**: Pide la longitud de la contraseña. La función seguirá funcionando hasta que el usuario escriba un número válido.

**Función**: se elige, una mayúscula, una minúscula, un número y un carácter especial aleatorios. Así nos aseguramos que haya un carácter de cada uno.
Se complementa estos cuatro carácteres con el resto hasta conseguir la longitud que el usuario a elegido.
Se mezclan los elementos de la lista, une todos los elementos de la lista.

**Argumentos**: longitud de la contraseña

**Return**: Contraseña aleatoria.

```python
#Generar la contraseña
def generar_password(longitud):

    mayus = random.choice(string.ascii_uppercase)
    minus = random.choice(string.ascii_lowercase)
    numero = random.choice(string.digits)
    simbolo = random.choice(string.punctuation)

    caracteres = (
        string.ascii_letters
        + string.digits
        + string.punctuation
    )

    password = [mayus, minus, numero, simbolo]

    for i in range(longitud - 4):
        password.append(random.choice(caracteres))

    random.shuffle(password)

    return "".join(password)
```

```python
generar_password(8)
```

**Argumentos**: longitud.

**Return**: imprime la contraseña.

```python
def main_password():

    longitud = num_caracteres()
    contrasena = generar_password(longitud)

    print("Contraseña generada:", contrasena)
```

**Función**: Crea registro del equipo si no esta aún.

**Argumentos**: equipos y nombre

**Return**: Nos da el nombre del equipo y el resto de columnas a 0.

```python
def crear_equip(equips,nom):
  if nom not in equips:
    equips[nom] = {
        'gols_favor' : 0,
        'gols_contra' : 0,
        'guanyats' : 0,
        'empatats' : 0,
        'perduts' : 0,
        'punts' : 0
    }
```

**Función**: Suma los goles a favor o en contra del equipo local y visitante.

**Argumentos**: listado de equipos, equipo local, equipo visitante y goles.

**Return**: Nos da el nombre del equipo y el resto de columnas a 0.

```python
def actualitzar_gols(equips, equip1, equip2, gols1, gols2):

    equips[equip1]["gols_favor"] += gols1
    equips[equip1]["gols_contra"] += gols2

    equips[equip2]["gols_favor"] += gols2
    equips[equip2]["gols_contra"] += gols1
```

**Función**: Dependiendo quien haya ganado el partido suma los puntos correspondientes.

**Argumentos**: listado de equipos, equipo local, equipo visitante y goles.

**Return**: Puntos

```python
def actualitzar_resultat(equips, equip1, equip2, gols1, gols2):

    if gols1 > gols2:

        equips[equip1]["guanyats"] += 1
        equips[equip2]["perduts"] += 1
        equips[equip1]["punts"] += 3

    elif gols1 < gols2:

        equips[equip2]["guanyats"] += 1
        equips[equip1]["perduts"] += 1
        equips[equip2]["punts"] += 3

    else:

        equips[equip1]["empatats"] += 1
        equips[equip2]["empatats"] += 1

        equips[equip1]["punts"] += 1
        equips[equip2]["punts"] += 1
```

**Función**: procesa cada partido, separa los equipos del resultado. Después separa los goles. Comprueba si existe un equipo o no, en caso negativo, crea el equipo.

**Argumentos**: cada línea de partidos y los equipos

**Return**: añade goles, suma puntos, suma goles

```python
def processar_partit(linia, equips):

    equip1, resultat, equip2 = linia.strip().split("\t")

    gols1, gols2 = map(int, resultat.split("-"))

    crear_equip(equips, equip1)
    crear_equip(equips, equip2)

    actualitzar_gols(
        equips,
        equip1,
        equip2,
        gols1,
        gols2
    )

    actualitzar_resultat(
        equips,
        equip1,
        equip2,
        gols1,
        gols2
    )
```

**Función**: Ordena la clasificación de mayor a menor puntos obtenidos.

**Argumentos**: Equipos

**Return**: Listado ordenado

```python
def ordenar_classificacio(equips):

    return sorted(
        equips.items(),
        key=lambda x: x[1]["punts"],
        reverse=True
    )
```

**Función**: Da respuesta a qué equipo ha sido el mas goleador y el más goleado.

**Argumentos**: Equipos

**Return**: Imprime las respuestas

```python
def respuestas_liga(equips):

  mes_golejador = equips['gols_favor'].idxmax()
  goles1 = equips.loc[mes_golejador, 'gols_favor']
  print(f'\nEquip més golejador: {mes_golejador}, amb {goles1} goles\n')


  mes_golejat = equips['gols_contra'].idxmax()
  goles2 = equips.loc[mes_golejat, 'gols_contra']
  print(f'\nEquip que més gols ha rebut: {mes_golejat}, amb {goles2} goles\n')
```

**Función**: Crea un data frame con los equipos ordenador por los puntos de mayor a menor

**Argumentos**: Equipos

**Return**: Devuelve el dataframe

```python
def crear_dataframe(equips):

    df = pd.DataFrame.from_dict(
        equips,
        orient='index'
    )

    df.sort_values(
        by='punts',
        ascending=False,
        inplace=True
    )

    return df
```

**Función**: Junta todas las funciones anteriores.

**Argumentos**: Equipos

**Return**: Imprime la clasificación

```python
def main_lliga():

    ruta = obtener_ruta()

    equips = {}

    with open(ruta, "r", encoding="utf-8") as archivo:

        for linea in archivo:

            processar_partit(linea, equips)

    classificacio = ordenar_classificacio(equips)

    df = crear_dataframe(equips)

    respuestas_liga(df)

    print("\n========================")
    print("   CLASSIFICACIÓ FINAL")
    print("========================\n")

    print(df)
```
