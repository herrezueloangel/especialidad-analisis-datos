**``Km anuales``**

Sustituimos los valores extremos por NaN.

Chequeamos para valorar los kilómetros que están fuera de lo común.
Se aplica el percentil (posición relativa en la multitud):
* Percentil 25 — el valor por debajo del cual está el 25% de los datos.
* Percentil 75 — el valor por debajo del cual está el 75% de los datos.
* IQR --> El rango intercuartílico — la distancia entre el percentil 75 y el 25. Representa donde está el 50% central de los datos.
* Límite superior --> El límite a partir del cual un valor se considera outlier por arriba.
* Límite inferior --> Lo mismo pero por abajo.

```python
Q1 = df['Km_anuals'].quantile(0.25)
Q3 = df['Km_anuals'].quantile(0.75)
IQR = Q3 - Q1
limite_superior = Q3 + 1.5 * IQR
limite_inferior = Q1 - 1.5 * IQR
print(Q1, Q3, IQR, limite_superior, limite_inferior)
```

```python
df.loc[
    (df['Km_anuals']<0 ) | (df['Km_anuals'] > limite_superior),
 'Km_anuals'
 ] = np.nan
```

```python
print("Negativos:", (df['Km_anuals'] < 0).sum())
print("Superiores al límite:", (df['Km_anuals'] > limite_superior).sum())
```

```python
#convertimos los índices de coste de vida y transporte público en categorías para facilitar su análisis y visualización.
# Los rangos los he establecido yo.

def categorizar_cost_vida(valor):
    if pd.isna(valor):
        return np.nan
    elif valor < 65:
        return 'low'
    elif valor <= 80:
        return 'medium'
    else:
        return 'high'

def categorizar_transport(valor):
    if pd.isna(valor):
        return np.nan
    elif valor < 7.0:
        return 'low'
    elif valor <= 8.5:
        return 'medium'
    else:
        return 'high'

df['cost_vida_cat'] = df['Cost_vida_index'].apply(categorizar_cost_vida)
df['transport_cat'] = df['Transport_public_score'].apply(categorizar_transport)

print(df['cost_vida_cat'].value_counts(dropna=False))
print()
print(df['transport_cat'].value_counts(dropna=False))
```

```python
#Ahora preparamos las dos columnas de edad y de franja de edad.

df['Edat'] = 2025 - df['Any de Naixement'].astype(int)

def franja_edat(edat):
    if edat < 30:
        return '18-29'
    elif edat < 45:
        return '30-44'
    elif edat < 60:
        return '45-59'
    else:
        return '60+'

df['Franja_edat'] = df['Edat'].apply(franja_edat)

print(df['Franja_edat'].value_counts())
```

```python
# Calculamos el incremento para cada empleado según la política salarial.

def calcular_increment(row):

    # Incremos en base a grupo profesional
    grup = row['Grup Professional']
    inc_grup = politica['base_increment_by_group'].get(grup, 0)

    # ajuste por hijos
    te_fills = pd.notna(row['Nombre_fills']) and row['Nombre_fills'] > 0
    inc_fills = politica['children_adjustment']['has_children'] if te_fills else politica['children_adjustment']['no_children']

    # ajuste por edad
    inc_edat = politica['age_adjustment'].get(row['Franja_edat'], 0)

    # Ajusta por coste de vida (NaN si no hay datos)

    if pd.isna(row['cost_vida_cat']):
        inc_cost = np.nan
    else:
        inc_cost = politica['cost_of_living_adjustment'].get(row['cost_vida_cat'], 0)

    # Ajustar por transporte (NaN si no hay datos)
    if pd.isna(row['transport_cat']):
        inc_transport = np.nan
    else:
        inc_transport = politica['transport_adjustment'].get(row['transport_cat'], 0)

    # Total
    total = inc_grup + inc_fills + inc_edat
    if not pd.isna(inc_cost):
        total += inc_cost
    if not pd.isna(inc_transport):
        total += inc_transport

    # Aplicar tope máximo
    total = min(total, politica['rules']['max_total_increment_percent'])

    return pd.Series({
        'inc_grup': inc_grup,
        'inc_fills': inc_fills,
        'inc_edat': inc_edat,
        'inc_cost_vida': inc_cost,
        'inc_transport': inc_transport,
        'increment_total': total
    })

increments = df.apply(calcular_increment, axis=1)
df = pd.concat([df, increments], axis=1)

print(df['increment_total'].describe().round(2))
```

```python
df['Salari_ajustat'] = df['Salari_numeric'] * (1 + df['increment_total'] / 100)

df[['Nom', 'Ciutat', 'Salari_numeric', 'increment_total', 'Salari_ajustat']].head(10)
```

```python
# Por parte de la empresa nos dice que el mínimo salarío es de 800€, por lo aquellos salarios que estan por debajo de ese mínimo, lo ajustamos a 800€.
# .clip() es una función de pandas que permite recortar los valores de una columna a un rango específico, en este caso, establece el minimo en 800€.

df['Salari_ajustat'] = df['Salari_ajustat'].clip(lower=800)

# Verificamos
por_debajo_minimo = df[df['Salari_ajustat'] < 800]
print(f"Empleados por debajo de 800€ tras aplicar el mínimo: {len(por_debajo_minimo)}")
```

### **Exercici 3: Creació de variables derivades (Feature Engineering)**

* **Ràtio salari / cost de vida.** Nos ayudará a saber si el salario que percibe es suficiente en el contexto geográfico donde vive.
* **Indicador de dependència familiar:** Salari / Nombre fills. Nos indica a saber cuanto salario tiene disponible para cada hijo. cuantos más hijos con el mismo salario, menor capacidad económica.
* **Indicador de etapa vital:** 0 → sense càrregues, 1 → etapa activa, 2 → alta dependència.Nos ayuda a comparar salarios de empleados con diferentes cargas familiares. Nos podría ayudar a diseñar un programa de beneficios para personas con hijos.

```python
# Salario/coste de vida

df['ratio_salari_cost_vida'] = (df['Salari_numeric'] / df['Cost_vida_index']).round(2)

df[['Nom', 'Ciutat', 'Salari_numeric', 'Cost_vida_index', 'ratio_salari_cost_vida']].head(10)

# Como teníamos 121 empleados sin datos de ciudad, y por tanto sin datos de coste de vida, el ratio de salario sobre coste de vida no se puede calcular para esos empleados,
# por lo que tendremos 121 valores NaN en esa nueva columna.
```

### **Exercici 2:  Optimització de rutes (problema aplicat)**



```python
df_distancies = pd.read_excel("/content/drive/MyDrive/ANALISTA DE DATOS/ESPECIALIDAD/SPRINT 8/matriu_distancies.xlsx", index_col=0)

df_distancies.head()
```

```python
ciutats_eliminar = ['Palma', 'Las Palmas de Gran Canaria']

df_distancies = df_distancies.drop(
    index=ciutats_eliminar,
    columns=ciutats_eliminar
)

df_distancies.head()
```

```python
print(df_distancies.shape)
```

```python
def calcular_ruta(df, origen, desti):

    ruta = [origen]

    visitades = [origen]

    ciutat_actual = origen

    distancia_total = 0


    while ciutat_actual != desti:

        # Distancias desde la ciudad actual
        distancies = df.loc[ciutat_actual]

        # Eliminar ciudades visitadas
        distancies = distancies.drop(visitades)

        # Ciudad más cercana
        propera_ciutat = distancies.idxmin()

        # Distancia mínima
        distancia = distancies.min()

        # Actualizar ruta
        ruta.append(propera_ciutat)

        # Actualizar visitadas
        visitades.append(propera_ciutat)

        # Sumar distancia
        distancia_total += distancia

        # Moverse a la nueva ciudad
        ciutat_actual = propera_ciutat


    return ruta, distancia_total
```

```python
ruta, distancia = calcular_ruta(df_distancies, 'Barcelona', 'Vigo')

print(ruta)
print(distancia)
```
