### Importamos la librería request para poder hacer peticiones a una API

```python
import requests
import pandas as pd
```

**Nivell 1**

Se crea una variable con la dirección url que vamos a usar

```python
url_api = "https://jsonplaceholder.typicode.com"
```

Con .get hacemos la primera petición. En este caso, pedimos la lista de todas las publicaciones.

```python
#Lista de posts

respuesta_posts = requests.get(url_api + "/posts")
```

Comprobamos el código del estado, depende del número que salga puede darnos error, en este caso es 200, lo que significa que la peticion ha ido bien.

```python
respuesta_posts.status_code
```

Cuando hacemos una petición a una API, devuelve los datos en formato JSON. Para que Python trabaje mejor con los datos, convertimos la respuesta en objetivos de Python, así resultatá más fácil manipularlos.

```python
posts = respuesta_posts.json()
```

```python
print("Código de estado:", respuesta_posts.status_code)
print("Total de publicaciones:", len(posts))
```

```python
df_posts = pd.DataFrame(posts)

df_posts
```

Hacemos lo mismo para usuarios

```python
#Lista usuarios

respuesta_users = requests.get(url_api + "/users")

respuesta_users.status_code

users = respuesta_users.json()

print("Código de estado:", respuesta_users.status_code)
print("Total de usuarios:", len(users))
```

```python
df_users = pd.DataFrame(users)

df_users
```

```python
#Lista tareas

respuesta_todos = requests.get(url_api + "/todos")

respuesta_todos.status_code

todos = respuesta_todos.json()

print("Código de estado:", respuesta_todos.status_code)
print("Total de tareas:", len(todos))
```

```python
df_todos = pd.DataFrame(todos)

df_todos
```

Creamos una variable con un diccionario con las partes del post: título, cuerpo del post y el usuario que ha escrito el post.

```python
nueva_publicacion = {
    "title": "Mi primera publicación",
    "body": "Lorem ipsum dolor sit amet consectetur adipiscing elit donec ligula, rhoncus sapien vulputate laoreet eros hendrerit varius in.",
    "userId": 1
}
```

### PATCH

Creamos una variable con un diccionario con los cambios que queremos realizar.

```python
cambios = {
    "title": "Título modificado"
}
```

Usamos patch para hacer el cambio solo en la parte que queremos. En la url especificamos que post es exactamente.

```python
cambios_post = requests.patch(
    url_api + "/posts/1",
    json=cambios
)
```

Verificamos que la respuesta es ok

```python
cambios_post.status_code
```

Y vemos que se realiazó el post correctamente.

```python
cambios_post.json()
```

### DELETE

Para borrar un post debemos usar el .delete especificando en la url el post que queremos eliminar.

```python
borrar_post = requests.delete(
    url_api + "/posts/1"
)
```

```python
print("Código de estado:", borrar_post.status_code)
print("Respuesta:", borrar_post.json())
```

## API seleccionada: MeowFacts

URL base:
https://meowfacts.herokuapp.com/

Dentro del github de la API meowfact > docs > api.jason, podemos ver un apartado dentro de "path{" con los endpoints:

- `/` → devuelve una o varias curiosidades sobre gatos.
- `/health` → comprueba si la API está activa.
- `/options` → devuelve las opciones disponibles, como los idiomas.

Tambiém podemos encontrar los parámetros:

- `factID` → permite pedir una curiosidad concreta.
- `count` → número de curiosidades que queremos recibir.
- `lang` → idioma de la respuesta.

en el apartado de responses vemos que las respuestas seran en formato json:

"application/json"

Se guarda la url en una variable, hacemos la petición con .get con dos parámetros dentro de la petición, que nos de cinco datos sobre los gatos ("counts" :5) y que sea en inglés ("lang":"eng").

Se imprime el código de estado de la petición: 200,petición funciona correctamente.

```python
#GET

url_meow = "https://meowfacts.herokuapp.com/"

respuesta_meow = requests.get(
    url_meow,
    params = {
        "count":5,
        "lang" : "eng"} 
)

print("Código de estado:", respuesta_meow.status_code)
```

Se convierte la respuesta que da la API a objetos de Python.
datos_meow es un diccionario, data seria la clave y la o las frases en una lista dentro de esa clave.

```python
datos_meow = respuesta_meow.json()
```

```python
datos_meow["data"]
```

Se convierte los datos en un DataFrame.

datos_meow["data"] -> lista de datos de los gatos.

columns=["fact"] -> nombre de la columna

```python
df_meow = pd.DataFrame(datos_meow["data"], columns=["fact"])
```

Como Python me mostraba el texto cortado, pasamos el DataFRame a cadena de texto

```python
print(df_meow.to_string())
```

```python
for dato in df_meow["fact"]:
    print(dato)
    print()
```

1. Utilitza l’API d’Open Data BCN per:

Cercar un dataset del teu interès mitjançant package_search i package_show. .

```python
url_bcn = "https://opendata-ajuntament.barcelona.cat/data/api/3/action"
```

He guardado en una variable una lista de temas que me podían interesar.

Con un bucle for he hecho la petición de que hiciese una búsqueda con los parámetros que había guardado en la variable.

Convierte la respuesta de la API en objeto de Python.
y mostramos por cada parámetro de la lista los 5 primeros temas.

```python
busquedas = ["poblacio", "habitatge", "accidents"]

for busqueda in busquedas:
    respuesta = requests.get(
        url_bcn + "/package_search",
        params={"q": busqueda}
    )

    print(f"\nResultados para: {busqueda}")
    
    datos = respuesta.json()

    for dataset in datos["result"]["results"][:5]:
        print(dataset["title"])
```

Después de revisar los temas, he elegido habitatge, asi que hago una nueva petición solo con los datasets de habitatge.
Se convierten en objeto Python.

```python
respuesta_search = requests.get(
    url_bcn + "/package_search",
    params={"q": "habitatge"}
)

print(respuesta_search.status_code)

datos_search = respuesta_search.json()
```

Miramos la estructura dataset

```python
print(datos_search.keys())
```

```python
print(datos_search["result"].keys())
```

Por cada dataset que hay dentro de la lista datos_search["result"]["results"], imprime su título, nombre y una linea separadora.

```python
for dataset in datos_search["result"]["results"]:
    print(dataset["title"])
    print(dataset["name"])
    print("-" * 50)
```

2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

Se hace una petición a la API para obtener toda la información del dataset escogido: 

notaris_compravendes_trans_tip_us

```python
respuesta_show = requests.get(
    url_bcn + "/package_show",
    params={"id": "notaris_compravendes_trans_tip_us"}
)

print(respuesta_show.status_code)

datos_show = respuesta_show.json()
```

Para cada recurso que hay dentro, imprime su nombre, el formato del archivo y el ID.

3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:

Recuperar almenys 100 registres

```python
#Recurso seleccionado 2025

resource_id = "4b8c5cf6-56aa-4a64-b87f-6159a1272206"
```

Hacemos una petición con /datastore_search, para que nos dé las filas con los datos.
