```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path(
    r"C:\Users\herre\OneDrive\Desktop\Proyectos DATA\SPRINT 13"
)

RAW_POBLACION = (
    PROJECT_ROOT
    / "01_data"
    / "01_raw"
    / "indicadores sociales"
    / "poblacion"
)

PROCESSED_POBLACION = (
    PROJECT_ROOT
    / "01_data"
    / "02_processed"
    / "indicadores_sociales"
    / "poblacion"
)

PROCESSED_POBLACION.mkdir(parents=True, exist_ok=True)
```

```python
ruta_entrada = RAW_POBLACION / "56940.csv"

poblacion = pd.read_csv(
    ruta_entrada,
    sep=";",
    encoding="latin1"
)

poblacion.head()
```

```python
poblacion_filtrada = poblacion[
    (poblacion['Periodo'].str.startswith('1 de enero')) &
    (poblacion['Comunidades y ciudades autónomas'] != 'Total Nacional')
].copy()

poblacion_filtrada.head()
```

```python
poblacion_filtrada.shape
```

```python
poblacion_filtrada['comunidad'] = (
    poblacion_filtrada['Comunidades y ciudades autónomas']
    .str.replace(r'^\d+\s+', '', regex=True)
)

poblacion_filtrada[['Comunidades y ciudades autónomas', 'comunidad']].drop_duplicates()
```

```python
mapa_comunidades = {
    'Asturias, Principado de': 'Asturias',
    'Balears, Illes': 'Baleares',
    'Castilla - La Mancha': 'Castilla-La Mancha',
    'Comunitat Valenciana': 'Comunidad Valenciana',
    'Madrid, Comunidad de': 'Madrid',
    'Murcia, Región de': 'Murcia',
    'Navarra, Comunidad Foral de': 'Navarra',
    'Rioja, La': 'La Rioja'
}

poblacion_filtrada['comunidad'] = (
    poblacion_filtrada['comunidad']
    .replace(mapa_comunidades)
)
```

```python
poblacion_filtrada['comunidad'].unique()
```

```python
poblacion_filtrada['año'] = (
    poblacion_filtrada['Periodo']
    .str.extract(r'(\d{4})')
    .astype(int)
)

poblacion_filtrada[['Periodo', 'año']].head()
```

```python
poblacion_final = poblacion_filtrada[
    ['comunidad', 'año', 'Total']
].copy()

poblacion_final = poblacion_final.rename(
    columns={'Total': 'poblacion'}
)

poblacion_final.head()
```

```python
poblacion_final.info()
```

```python
poblacion_final['poblacion'] = (
    poblacion_final['poblacion']
    .str.replace('.', '', regex=False)
    .astype(int)
)
```

```python
poblacion_final.describe()
```

```python
ruta_salida = PROCESSED_POBLACION / "poblacion.csv"

poblacion_final.to_csv(
    ruta_salida,
    index=False
)

print(f"Archivo guardado en: {ruta_salida}")
```
