```python
def extraer_comentarios_video(youtube, video_id, max_comentarios=1000):

    comentarios = []
    page_token = None

    while len(comentarios) < max_comentarios:

        request = youtube.commentThreads().list(
            part="snippet",
            videoId=video_id,
            maxResults=100,
            textFormat="plainText",
            pageToken=page_token,
            order="relevance"
        )

        response = request.execute()

        for item in response["items"]:

            comentario = item["snippet"]["topLevelComment"]["snippet"]

            comentarios.append({
                "video_id": video_id,
                "autor": comentario["authorDisplayName"],
                "fecha": comentario["publishedAt"],
                "likes": comentario["likeCount"],
                "texto": comentario["textDisplay"]
            })

            if len(comentarios) >= max_comentarios:
                break

        page_token = response.get("nextPageToken")

        if not page_token:
            break

    return pd.DataFrame(comentarios)
```

## 1. Obtener estadísticas

```python
video_ids = df_videos_nuevos_reales_2019["video_id"].tolist()

estadisticas_nuevas = []

for i in range(0, len(video_ids), 50):

    lote_ids = video_ids[i:i + 50]

    request_stats = youtube.videos().list(
        part="statistics",
        id=",".join(lote_ids)
    )

    response_stats = request_stats.execute()

    for item in response_stats["items"]:

        stats = item.get("statistics", {})

        estadisticas_nuevas.append({
            "video_id": item["id"],
            "visualizaciones": int(stats.get("viewCount", 0)),
            "likes": int(stats.get("likeCount", 0)),
            "comentarios": int(stats.get("commentCount", 0))
        })

df_stats_nuevos_2019 = pd.DataFrame(estadisticas_nuevas)
```

## 2. Unir estadísticas

```python
df_videos_nuevos_reales_2019 = (
    df_videos_nuevos_reales_2019
    .merge(
        df_stats_nuevos_2019,
        on="video_id",
        how="left"
    )
)
```

## 3. Seleccionar los 100 con más comentarios

```python
df_seleccion_nuevos_2019 = (
    df_videos_nuevos_reales_2019
    .sort_values(
        "comentarios",
        ascending=False
    )
    .head(100)
    .copy()
)

df_seleccion_nuevos_2019[
    ["canal", "titulo", "comentarios", "visualizaciones"]
].head(30)
```
