```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
RUTA = r"C:\Users\herre\OneDrive\Desktop\Proyectos DATA\SPRINT 13\01_data\02_processed\delito_de_odio"

desconocido = pd.read_csv(
    RUTA + r"\delitos_desconocido.csv",
    encoding="utf-8-sig"
)

extranjero = pd.read_csv(
    RUTA + r"\delitos_extranjero.csv",
    encoding="utf-8-sig"
)

nacional = pd.read_csv(
    RUTA + r"\delitos_nacional.csv",
    encoding="utf-8-sig"
)
```

```python
# Aplicar el mismo filtro que en el notebook 02
# Total calificación + Total general = una fila por ámbito y año
desconocido_limpio = desconocido[
    (desconocido['Calificación'] == 'TOTAL calificación') &
    (desconocido['Tipología penal'] == 'Total general')
].copy()

extranjero_limpio = extranjero[
    (extranjero['Calificación'] == 'TOTAL calificación') &
    (extranjero['Tipología penal'] == 'Total general')
].copy()

nacional_limpio = nacional[
    (nacional['Calificación'] == 'TOTAL calificación') &
    (nacional['Tipología penal'] == 'Total general')
].copy()

print("Desconocido:", desconocido_limpio.shape)
print("Extranjero:", extranjero_limpio.shape)
print("Nacional:", nacional_limpio.shape)
```

```python
# Filtrar solo TOTAL ámbitos para ver evolución general
desc_total = desconocido_limpio[
    desconocido_limpio['Ámbito'] == 'TOTAL ámbitos'
].copy()

ext_total = extranjero_limpio[
    extranjero_limpio['Ámbito'] == 'TOTAL ámbitos'
].copy()

nac_total = nacional_limpio[
    nacional_limpio['Ámbito'] == 'TOTAL ámbitos'
].copy()

# Combinar los tres para comparar
comparacion = pd.DataFrame({
    'año': desc_total['periodo'].values,
    'Desconocido': desc_total['Total'].values,
    'Extranjero': ext_total['Total'].values,
    'Nacional': nac_total['Total'].values
})

print(comparacion)
```
