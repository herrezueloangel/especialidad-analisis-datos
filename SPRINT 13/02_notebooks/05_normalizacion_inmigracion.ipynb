```python
import pandas as pd
from pathlib import Path
```

```python
ruta = r"C:\Users\herre\OneDrive\Desktop\Proyectos DATA\SPRINT 13\01_data\01_raw\indicadores sociales\inmigracion\79277.csv"

df = pd.read_csv(
    ruta,
    encoding='latin-1',
    sep=None,
    engine='python'
)

print(df.shape)
df.head()
```

```python
# Filtrar solo el dato de 1 de enero de cada año
df = df[df['Periodo'].str.startswith('1 de enero')].copy()

# Extraer el año del periodo
df['año'] = df['Periodo'].str.extract(r'(\d{4})').astype(int)

# Limpiar Total: quitar puntos de miles y convertir a número
df['Total'] = (
    df['Total'].astype(str)
    .str.replace('.', '', regex=False)
    .pipe(pd.to_numeric, errors='coerce')
)

print(df.shape)
print(sorted(df['año'].unique()))
df.head()
```

```python
# Quitar 2025 (no lo necesitamos)
df = df[df['año'] <= 2024].copy()

# Quedarnos solo con columnas útiles y renombrar
df = df.rename(columns={
    'Comunidades y ciudades autónomas': 'comunidad',
    'Total': 'inmigrantes'
})[['comunidad', 'año', 'inmigrantes']].copy()

# Limpiar prefijo numérico y nombres largos
df['comunidad'] = (
    df['comunidad']
    .str.replace(r'^\d+\s+', '', regex=True)
    .str.strip()
    .str.replace('Asturias, Principado de', 'Asturias', regex=False)
    .str.replace('Balears, Illes', 'Baleares', regex=False)
    .str.replace('Castilla - La Mancha', 'Castilla-La Mancha', regex=False)
    .str.replace('Comunitat Valenciana', 'Comunidad Valenciana', regex=False)
    .str.replace('Madrid, Comunidad de', 'Madrid', regex=False)
    .str.replace('Murcia, Región de', 'Murcia', regex=False)
    .str.replace('Navarra, Comunidad Foral de', 'Navarra', regex=False)
    .str.replace('Rioja, La', 'La Rioja', regex=False)
)

print(df.shape)
print(sorted(df['comunidad'].unique()))
df.head()
```

```python
import os

RUTA_PROCESADOS = r"C:\Users\herre\OneDrive\Desktop\Proyectos DATA\SPRINT 13\01_data\02_processed\indicadores_sociales\inmigracion"
os.makedirs(RUTA_PROCESADOS, exist_ok=True)

df.to_csv(
    RUTA_PROCESADOS + r"\inmigracion_españa.csv",
    index=False,
    encoding='utf-8-sig'
)

print(f"Guardado correctamente: {len(df):,} filas")
```
